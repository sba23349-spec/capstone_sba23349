```python
from pathlib import Path

files = sorted(Path(".").glob("*.csv"))

total = 0

for file in files:
    with open(file, encoding="utf-8-sig") as f:
        rows = sum(1 for _ in f) - 1

    print(f"{file.name}: {rows:,} rows")
    total += rows

print(f"\nTotal: {total:,} rows")
print(f"Files found: {len(files)}")
```
